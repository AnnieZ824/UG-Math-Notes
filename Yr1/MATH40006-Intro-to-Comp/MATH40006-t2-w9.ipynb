## 1 Grouping data

### 1. .groupby

### 2. .resample(): ".groupby" according to $\textbf{time}$
    1. Datetime as index
    2. 'ME': month end
    3. + functions: .last() / .mean()

### 3. WITH .iloc[] / .loc[]: filter data
    1. .iloc[]: locate by index
    2. .loc[]: locate by name, 'col','row'

### 4. WITH .agg(): combine functions
    1. aggregate different functions from .resample or .groupby
    2. use {} to include a list, '' to include name of the function

```python
# Example

# read csv
df = pd.read_csv('historical_stock_market.csv')
stocks_df2 = df.copy()
stocks_df2['Date'] = pd.to_datetime(stocks_df2['Date'])
stocks_df2.set_index('Date', inplace=True)

# resample by time
stock_resampled = stocks_df2.iloc[:-4].resample('ME').agg({
    'Open': 'first',
    'Close': 'last'
    })
stock_resampled.head()
```
